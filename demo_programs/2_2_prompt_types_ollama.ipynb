```python
import ollama
```

```python
one_shot_prompt = (
    f"Example:\n"
    f"Q: Translate 'Hello, how are you?' into Hindi.\n"
    f"A: Aap kaise ho?\n\n"
    f"Now answer this:\n"
    f"Q: Translate 'What is your name?' into Hindi.\nA:"
)

few_shot_prompt = """
Example:
Message: My laptop hangs frequently
Category: Laptop

Message: Food quality is poor in canteen
Category: Food

Now classify the following message:
Message: Bus driver is a rash driver
Category:
"""
response = ollama.chat(
    model="gpt-oss:120b-cloud",
    messages=[
        {"role": "system", "content": "You are an admin officer"},
        {"role": "user", "content": few_shot_prompt},
        {"role": "assistant", "content": "Please answer in one word"}
        
    ]
)
print(response['message']['content'])
```
